Notebook ini melakukan EDA awal pada dataset retail untuk memvalidasi fitur RFM sebelum clustering.

```python
import sys
sys.path.append('..')

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from src.preprocessing import clean_transactions
from src.rfm import build_rfm

sns.set_style('whitegrid')
pd.set_option('display.max_columns', None)
```

## 1. Load Data

```python
df = pd.read_csv('../data/sample_data.csv')
print('Shape:', df.shape)
df.head()
```

```python
df.info()
df.isnull().sum()
```

## 2. Data Cleaning

```python
df_clean = clean_transactions(df)
print('Before:', df.shape, '| After:', df_clean.shape)
df_clean.head()
```

## 3. Statistik Deskriptif

```python
df_clean.describe()
```

```python
print('Total revenue:', df_clean['TotalPrice'].sum())
print('Unique customers:', df_clean['CustomerID'].nunique())
print('Unique products:', df_clean['StockCode'].nunique())
print('Date range:', df_clean['InvoiceDate'].min(), '->', df_clean['InvoiceDate'].max())
```

## 4. Top Products

```python
top_products = (
    df_clean.groupby(['StockCode', 'Description'])
    .agg(qty=('Quantity', 'sum'), revenue=('TotalPrice', 'sum'))
    .sort_values('revenue', ascending=False)
    .head(10)
)
top_products
```

## 5. RFM Feature Engineering

```python
rfm = build_rfm(df_clean)
rfm.describe()
```

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
for ax, col in zip(axes, ['Recency', 'Frequency', 'Monetary']):
    sns.histplot(rfm[col], ax=ax, kde=True)
    ax.set_title(f'Distribusi {col}')
plt.tight_layout()
plt.show()
```
